```python
import matplotlib.pyplot as plt
import numpy as np
x = np.arange(0,10,1)
y = -2*(x-3) + 10


fig = plt.figure()
plt.plot(x,y)
fig.set_facecolor("#A5A4A4")
ax = plt.gca()
plt.title("Demand Curve of Product A")
plt.hlines(8, 0, 4, linestyles=(0,(1,2)), colors="black")
plt.vlines(4, 0, 8, linestyles=(0,(1,2)), colors="black")
plt.plot(4,8, "o")
plt.hlines(4, 0, 6, linestyles=(0,(1,2)), colors="black")
plt.vlines(6, 0, 4, linestyles=(0,(1,2)), colors="black")
plt.plot(6,4, "o")
plt.annotate("($P_1$, $Q_1$)", (4.2,8))
plt.annotate("($P_2$, $Q_2$)", (6.2, 4))
plt.annotate("", (3.5,8), (3.5, 4), arrowprops=dict(arrowstyle='<->'))
ax.set_xlabel("Quantity Demanded")
ax.set_ylabel("Unit Price")
plt.rc("font", size=10)
plt.text(3.2, 6, "$\\Delta P$", rotation=90)
plt.text(5, 2.6, "$\\Delta Q_d$")
plt.annotate("", (4,3), (6, 3), arrowprops=dict(arrowstyle='<->'))
ax.set_xlim([0,10])
ax.set_ylim([0,10])
ax.set_xticks([])
ax.set_yticks([])
ax.set_xticklabels([])
ax.set_yticklabels([])
# plt.savefig("graphs/demand-curve-price-change.png")
plt.show()
```

```python
## Graph for supply curve

plt.close('all')

xc = np.arange(0,10,1)
yc = 2*(x+3) - 10

def y(x):
    return 2*(x+3)-10

fig = plt.figure()
fig.set_facecolor("#A5A4A4")
plt.plot(xc,yc)
plt.title("Supply Curve of Product A")

axes = plt.gca()
axes.set_xlim([0,10])
axes.set_ylim([0,10])
axes.set_ylabel("Unit Price")
axes.set_xlabel("Quantity Supplied")
axes.set_xticks([])
axes.set_yticks([])
plt.hlines([y(4), y(6)], [0],[4,6], linestyles=(0,(1,2)), colors="black")
plt.vlines([4, 6],[0],[y(4), y(6)], linestyles=(0,(1,2)), colors="black")
plt.annotate("", (4.1, y(4)), (6, y(4)), arrowprops=dict(arrowstyle="<->"))
plt.annotate("", (4, y(4)+0.1), (4, y(6)), arrowprops=dict(arrowstyle="<->"))
plt.text(3.7, (y(4)+y(6))/2, "$\\Delta P$", rotation=90)
plt.text(4.5, y(4)-0.4, "$\\Delta Q_s$")
plt.plot(4,y(4), "o")
plt.plot(6, y(6), "o")
# plt.savefig("graphs/price-change-s-curve.png")

```

```python
def f(x):
    return 2*(x+3)-10

def g(x):
    return 2*(x+1)-10

def h(x):
    return 2*(x+5)-10   

x = [i for i in range(10)]
y = [f(i) for i in x]
y2 = [g(i) for i in x]
y3 = [h(i) for i in x]


fig = plt.figure()
fig.set_facecolor("#A5A4A4")
plt.rc("font", size=8)

axes = plt.gca()
axes.set_xlim([0,10])
axes.set_ylim([0,10])
axes.set_xticks([])
axes.set_yticks([])
axes.set_xlabel("Quantity supplied")
axes.set_ylabel("Unit Price")

plt.title("Supply Curve of Product A")
plt.plot(x,y)
plt.plot(x,y2, linestyle="dotted")
plt.plot(x,y3, linestyle="dotted")
plt.plot([2,4,6], [f(4), f(4), f(4)], "kx")
plt.vlines([2,4,6], [0,0,0], [f(4), f(4), f(4)], linestyles="dotted")
plt.hlines([f(4)], 0, 6, linestyles="dotted")
plt.rc("font", size=10)
plt.text(2.3, f(4.1), "$Q_{s1}$")
plt.text(4.3, f(4.1), "$Q_{s2}$")
plt.text(6.3, f(4.1), "$Q_{s2}$")
# plt.text(0.3, f(4.1), "$P_1$")
# axes.set_yticklabels("$P_1$")
axes.set_yticks([f(4)], ["$P_1$"])


aStyle = dict(arrowstyle="<-")
plt.annotate("", (6, f(6)), (8, f(6)), arrowprops=aStyle)
plt.annotate("", (6, f(6)), (4, f(6)), arrowprops=aStyle)
# plt.savefig("graphs/decrease-in-supply")

```
